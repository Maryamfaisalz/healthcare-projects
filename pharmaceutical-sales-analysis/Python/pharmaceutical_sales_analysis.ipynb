Q1 : What are the total sales quantities for each drug category (ATC code)?

- N05B is the second-largest contributor 14.61%

Q6 : Are respiratory drugs (R03) sold more during specific months?

```python
#A lookup dictionary mapping month numbers(1-12) to month names, useful for labeling monthly analysis and charts
month_names = {
    1:'January',
    2:'February',
    3:'March',
    4:'April',
    5:'May',
    6:'June',
    7:'July',
    8:'August',
    9:'September',
    10:'October',
    11:'November',
    12:'December'
}
```

```python
#Defines the calendar order of month names so monthly results can be displayed chronologically
month_order = ['January','February','March','April','May','June','July','August', 'September','October','November','December']
```

```python
#Maps month numbers to month names and stores them in a new month_names column
df["Month_names"] = df['Month'].map(month_names)
```

```python
#Average Monthly R03 Sales - calulates each month's mean R03 sales, then ranks months from highest to lowest
monthly_r03_sales = df.groupby("Month_names")["R03"].mean().reindex(month_order)
monthly_r03_sales_sorted = monthly_r03_sales.sort_values(ascending=False)
```

```python
#A labeled bar chart showing average daily R03 sales for each month, calendar order 
plt.figure(figsize=(18,8)) 
plt.title("Average daily sales by month")
plt.xlabel("Month Name")
plt.ylabel("Average Daily Sales Quantity")
bar_2 =plt.bar(monthly_r03_sales.index, monthly_r03_sales.values)
label = [round(value,1) for value in monthly_r03_sales.values]
plt.bar_label(bar_2,label)
plt.show()
```

Season in Serbia (according to the dataset):

- Winter: Dec - Feb

- Spring: March - May

- Summer: June - August

- Autumn: September - November


Business Insights

- Winter (Dec - Feb) have higher R03 sales 
- Summber (Jun - Aug) have lower R03 sales indicating a possible seasonal variation in demand for respiratory medicines
- The pharmacy could consider increasing stock levels ahead of winter to plan for potentially higher demand and reduce the risk of stock shortages.
- Historical monthly sales trend can be used to predict future demand and support purchasing and replenishment decisions
